<H1>ResearchVault</H1>

<H2>Problem</h2 >

</H3>Students and researchers often need to read multiple research papers to find specific information. Reading large documents manually is slow and inefficient.

The goal of ResearchVault is to allow users to upload research papers or PDF documents and ask questions about their contents.</H3>

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_community.vectorstores import FAISS
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import Docx2txtLoader
from langchain_core.prompts import ChatPromptTemplate
```

<H2>DOCUMENT LOADING</H2>

```python
loader = Docx2txtLoader('Ideology.docx')
document = loader.load()
```

<h2>TEXT SPLITTING AND CHUNKING</h2>

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size =500,
    chunk_overlap = 80
)
```

```python
chunk = text_splitter.split_documents(document)
```

```python
print('the number of chunks is: ',len(chunk))
```

<h2>EMBEDDING</h2>

```python
embedding = HuggingFaceEmbeddings(
    model = 'sentence-transformers/all-MiniLM-L6-v2'
)
```

<h3>Creating Vector DataBase</h3>

```python
vector_db = FAISS.from_documents(
    chunk,
    embedding
)
```

<h3>CREATING RETRIVER</H3>

```python
retriver = vector_db.as_retriever(
    search_kwargs = {'k':5}
)
```

<h3>MODEL</h3>

```python
model = ChatGoogleGenerativeAI(
    model = 'gemini-3.6-flash',
    temperature = 0
)
```

<h3>PROMPT</h3>

```python
prompt = ChatPromptTemplate.from_template(
    '''
    Answer from the context. 
    if the question is out of context just say: "QUESTION OUT OF CONTEXT"

    Context
    {context}

    Question
    {question}
'''
)
```

<h3>ASK QUERY</H3>

```python
question = input('enter your question: ')
```

<h3>RETRIVE RELEVANT DATA</h3>

```python
retrivered_data = retriver.invoke(question)
```

<h3>CREATE THE CONTEXT</h3>

```python
context = '\n\n'.join(
    pages.page_content
    for pages in retrivered_data
)
```

<h3>CALL THE PROMPT</h3>

```python
final_prompt = prompt.invoke({
    'context':context,
    'question':question
})
```

<h3>CALL THE MODEL</h3>

```python
response = model.invoke(final_prompt)
```

<h3>PRINT THE RESULT</h3>

```python
print(response.text())
```
